# Web Performance Optimizations
This notebook inspects render-blocking assets, demonstrates async/defer loading, shows how to split critical CSS, applies lazy loading for below-the-fold content, and converts images to modern formats with responsive `srcset`.

```python
from pathlib import Path
from PIL import Image
from bs4 import BeautifulSoup

workspace = Path('c:/laragon/www/app')
public_dir = workspace / 'public'
index_html = workspace / 'index.html'

print('Workspace:', workspace)
print('Public dir exists:', public_dir.exists())
print('Index HTML exists:', index_html.exists())
```

## Analyze Render-Blocking Scripts with `async`/`defer`
Inspect the HTML to identify blocking script tags and show how to add `defer` or `async` to non-critical scripts.

```python
html = index_html.read_text(encoding='utf-8')
soup = BeautifulSoup(html, 'html.parser')
for script in soup.find_all('script'):
    print('script:', script.attrs.get('src'), 'type:', script.attrs.get('type'), 'defer:', script.attrs.get('defer'), 'async:', script.attrs.get('async'))

# Example transformation
for script in soup.find_all('script'):
    if script.attrs.get('type') == 'module' and script.attrs.get('src'):
        script['defer'] = ''

print('\nTransformed head script:')
print(soup.find('script'))
```
